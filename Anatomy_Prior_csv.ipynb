```python
import nibabel as nib
import numpy as np

def has_organ(zf, pid, organ='liver'):
    path = f'{pid}/segmentations/{organ}.nii.gz'
    zf.extract(path, path='./check')
    img = nib.load(f'./check/{path}')
    data = img.get_fdata()
    voxel_count = np.sum(data > 0)
    return voxel_count, voxel_count > 0
```

```python
results = []
with RemoteZip(url) as zf:
    for pid, stype in df[['image_id','study_type']].values[:20]:
        voxels, present = has_organ(zf, pid, 'liver')
        results.append((pid, stype, voxels, present))

for r in results:
    print(r)
```

```python
import nibabel as nib
import numpy as np
import pandas as pd
from remotezip import RemoteZip

url = "https://zenodo.org/records/10047292/files/Totalsegmentator_dataset_v201.zip?download=1"

organs = [
    "liver",
    "spleen",
    "kidney_left",
    "kidney_right",
    "pancreas",
    "stomach",
    "urinary_bladder"
]

results = []

with RemoteZip(url) as zf:
    for pid in normal_df["image_id"]:
        row = {"image_id": pid}
        for organ in organs:
            path = f"{pid}/segmentations/{organ}.nii.gz"
            try:
                zf.extract(path, path="./check")
                img = nib.load(f"./check/{path}")
                voxel_count = int(np.sum(img.get_fdata() > 0))
                row[organ] = voxel_count
            except Exception as e:
                row[organ] = -1  # file missing or unreadable
        results.append(row)

coverage_df = pd.DataFrame(results)
coverage_df.to_csv("organ_coverage_check.csv", index=False)
coverage_df.head()
```

```python
complete_cases = coverage_df[(coverage_df[organs] > 0).all(axis=1)]
print("Cases with all 7 organs present:", len(complete_cases))
```

```python
import os
import shutil
from google.colab import drive

# Mount Google Drive (will prompt for auth the first time)
drive.mount('/content/drive')

download_ids = complete_cases["image_id"].tolist()
save_root = "./data"
drive_backup_root = "/content/drive/MyDrive/organ_segmentation_backup"
os.makedirs(drive_backup_root, exist_ok=True)

organs = [
    "liver",
    "spleen",
    "kidney_left",
    "kidney_right",
    "pancreas",
    "stomach",
    "urinary_bladder"
]

with RemoteZip(url) as zf:
    for pid in download_ids:
        patient_dir = f"{save_root}/{pid}"
        drive_patient_dir = f"{drive_backup_root}/{pid}"

        # skip if already backed up (lets you safely re-run after a crash)
        if os.path.exists(f"{drive_patient_dir}/segmentations/urinary_bladder.nii.gz"):
            print(f"Skipping {pid} (already backed up)")
            continue

        os.makedirs(f"{patient_dir}/segmentations", exist_ok=True)

        # download CT scan
        ct_path = f"{pid}/ct.nii.gz"
        zf.extract(ct_path, path=save_root)

        # download each organ mask
        for organ in organs:
            mask_path = f"{pid}/segmentations/{organ}.nii.gz"
            zf.extract(mask_path, path=save_root)

        # back up this patient's folder to Google Drive
        shutil.copytree(patient_dir, drive_patient_dir, dirs_exist_ok=True)

        print(f"Done + backed up: {pid}")

print("All downloads complete.")
```
